```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# Скачиваем данные
data = pd.read_csv('realty_data.csv')

# Удаляем строки с пропущенными значениями
data = data.dropna(subset=['total_square', 'rooms', 'price'])

# Предобработка данных
features = data[['total_square', 'rooms']]  # Признаки
target = data['price']  # Целевая переменная

# Разделяем данные на обучающую и тестовую выборки
X_train, X_test, y_train, y_test = train_test_split(features, target, test_size=0.2, random_state=42)

# Обучаем модель
model = LinearRegression()
model.fit(X_train, y_train)

# Проверяем качество модели
print(f'Коэффициенты модели: {model.coef_}')
print(f'Перехват: {model.intercept_}')

```

```python
def predict_price(total_square, rooms):
    return model.predict([[total_square, rooms]])[0]
```
