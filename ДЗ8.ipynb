```python
import requests
import pandas as pd

# Получаем данные о книгах
url_books = "https://www.anapioficeandfire.com/api/books"
response_books = requests.get(url_books)
books_data = response_books.json()

# Сохраняем в DataFrame
books_df = pd.DataFrame(books_data)
print(books_df)
```

```python
# Получаем данные о домах
url_houses = "https://www.anapioficeandfire.com/api/houses"
response_houses = requests.get(url_houses)
houses_data = response_houses.json()

# Сохраняем в DataFrame
houses_df = pd.DataFrame(houses_data)
print(houses_df)
```

```python
# Получаем данные о домах с девизом
url_houses_with_motto = "https://www.anapioficeandfire.com/api/houses?hasWords=true"
response_houses_with_motto = requests.get(url_houses_with_motto)
houses_with_motto_data = response_houses_with_motto.json()

# Сохраняем в DataFrame
houses_with_motto_df = pd.DataFrame(houses_with_motto_data)
print(houses_with_motto_df)
```
